```python
# ===== Day 5: Core Feature - K-Means Customer Segmentation =====
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# ---- Load data directly from GitHub ----
URL = "https://raw.githubusercontent.com/varshayinithangaraju30-lgtm/customer-segmentation-kmeans/main/Mall_Customers.csv"
df = pd.read_csv(URL)
print("Data loaded:", df.shape)

# ---- Task 1: Core feature ----
def segment_customers(df, n_clusters=3):
    """Segment customers using K-Means. Returns df with Cluster column."""

    # ---- Task 2: Edge cases ----
    if df is None or df.empty:
        raise ValueError("Input data is empty")
    if n_clusters < 1:
        raise ValueError("n_clusters must be at least 1")
    if n_clusters > len(df):
        raise ValueError("n_clusters cannot be more than number of rows")

    df = df.copy().drop_duplicates().dropna()
    features = ["Age", "Annual Income (k$)", "Spending Score (1-100)"]
    X = StandardScaler().fit_transform(df[features])

    model = KMeans(n_clusters=n_clusters, init="k-means++",
                   n_init=10, random_state=42)
    df["Cluster"] = model.fit_predict(X)

    # Empty cluster check
    sizes = df["Cluster"].value_counts()
    if len(sizes) < n_clusters:
        print("Warning: some clusters are empty!")

    return df, model

result, model = segment_customers(df, n_clusters=3)
print(result.head(10))
print("\nCluster sizes:")
print(result["Cluster"].value_counts())

# ---- Task 3: Basic tests ----
def test_empty_data():
    try:
        segment_customers(pd.DataFrame(), 3)
        assert False, "Should have raised error"
    except ValueError:
        pass

def test_too_many_clusters():
    try:
        segment_customers(df, 1000)
        assert False, "Should have raised error"
    except ValueError:
        pass

def test_cluster_column_added():
    res, _ = segment_customers(df, 3)
    assert "Cluster" in res.columns

def test_cluster_count():
    res, _ = segment_customers(df, 3)
    assert res["Cluster"].nunique() == 3

def test_no_missing_clusters():
    res, _ = segment_customers(df, 2)
    assert res["Cluster"].isna().sum() == 0

test_empty_data()
test_too_many_clusters()
test_cluster_column_added()
test_cluster_count()
test_no_missing_clusters()
print("\nAll 5 tests passed ✅")
```
