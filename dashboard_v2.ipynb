```python
import pandas as pd
import plotly.express as px
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

URL = "https://raw.githubusercontent.com/varshayinithangaraju30-lgtm/customer-segmentation-kmeans/main/Mall_Customers.csv"
FEATURES = ["Age", "Annual Income (k$)", "Spending Score (1-100)"]

df = pd.read_csv(URL).drop_duplicates().dropna(subset=FEATURES)
X = StandardScaler().fit_transform(df[FEATURES])

model = KMeans(n_clusters=3, init="k-means++", n_init=10, random_state=42)
df["Cluster"] = model.fit_predict(X).astype(str)

fig = px.scatter(
    df,
    x="Annual Income (k$)",
    y="Spending Score (1-100)",
    color="Cluster",
    hover_data=["CustomerID", "Gender", "Age"],
    title="Customer Segments (K = 3)",
)
fig.update_traces(marker=dict(size=12))
fig.show()

print(df["Cluster"].value_counts().sort_index())
print(df.groupby("Cluster")[FEATURES].mean().round(1))
```
