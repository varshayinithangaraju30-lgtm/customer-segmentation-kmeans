```python
# ===== Day 4: Data Preparation Pipeline =====
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

# ---- Task 1: Load customer dataset ----
df = pd.read_csv("Mall_Customers.csv")
print("Shape:", df.shape)
print(df.head())

# ---- Task 2: Preprocess and normalize ----
def preprocess(df):
    df = df.copy()
    df = df.drop_duplicates()
    df = df.dropna()
    # Encode Gender (Male=0, Female=1)
    df["Gender"] = df["Gender"].map({"Male": 0, "Female": 1})
    features = ["Gender", "Age", "Annual Income (k$)", "Spending Score (1-100)"]
    X = df[features]
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)
    return df, X_scaled

# ---- Task 3: Validate inputs for K-Means ----
def validate(X_scaled):
    assert not np.isnan(X_scaled).any(), "Data has NaN values"
    assert np.isfinite(X_scaled).all(), "Data has infinite values"
    assert X_scaled.shape[0] > 2, "Need more rows than clusters"
    print("Validation passed: no NaN, all numeric, shape =", X_scaled.shape)

df_clean, X_scaled = preprocess(df)
validate(X_scaled)

print("\nScaled mean (should be ~0):", X_scaled.mean(axis=0).round(2))
print("Scaled std  (should be ~1):", X_scaled.std(axis=0).round(2))
```
