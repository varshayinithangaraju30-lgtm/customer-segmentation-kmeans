```python
import urllib.request

BASE = "https://raw.githubusercontent.com/varshayinithangaraju30-lgtm/customer-segmentation-kmeans/main/"
urllib.request.urlretrieve(BASE + "segmentation.py", "segmentation.py")

import segmentation as seg

URL = BASE + "Mall_Customers.csv"

df, X, model = seg.run_pipeline(URL, n_clusters=3)
print(df["Cluster"].value_counts())
print(seg.evaluate(X, df["Cluster"]))

X_2d, pca = seg.reduce_to_2d(X)
print("PCA shape:", X_2d.shape)

def test_pipeline_adds_cluster():
    assert "Cluster" in df.columns and df["Cluster"].nunique() == 3

def test_evaluate_keys():
    assert set(seg.evaluate(X, df["Cluster"])) == {"silhouette", "calinski_harabasz"}

def test_bad_k():
    try:
        seg.run_pipeline(URL, 1000)
        assert False
    except ValueError:
        pass

def test_pca_shape():
    assert X_2d.shape == (len(X), 2)

for t in [test_pipeline_adds_cluster, test_evaluate_keys, test_bad_k, test_pca_shape]:
    t()
    print("PASS:", t.__name__)
print("Refactored module works ✅")
```
