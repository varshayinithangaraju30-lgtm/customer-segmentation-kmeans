```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score, calinski_harabasz_score

URL = "https://raw.githubusercontent.com/varshayinithangaraju30-lgtm/customer-segmentation-kmeans/main/Mall_Customers.csv"
FEATURES = ["Age", "Annual Income (k$)", "Spending Score (1-100)"]

def load_and_scale(path):
    df = pd.read_csv(path)
    if df.empty:
        raise ValueError("Input data is empty")
    df = df.drop_duplicates().dropna(subset=FEATURES)
    X = StandardScaler().fit_transform(df[FEATURES])
    return df, X

def evaluate_clustering(X, labels):
    """Return silhouette score and Calinski-Harabasz index."""
    if len(set(labels)) < 2:
        raise ValueError("Need at least 2 clusters to evaluate")
    return {
        "silhouette": silhouette_score(X, labels),
        "calinski_harabasz": calinski_harabasz_score(X, labels),
    }

df, X = load_and_scale(URL)

results = []
for k in range(2, 8):
    model = KMeans(n_clusters=k, init="k-means++", n_init=10, random_state=42)
    labels = model.fit_predict(X)
    scores = evaluate_clustering(X, labels)
    results.append({
        "K": k,
        "Silhouette": round(scores["silhouette"], 3),
        "Calinski-Harabasz": round(scores["calinski_harabasz"], 1),
    })

results_df = pd.DataFrame(results)
print(results_df)

best_k = results_df.loc[results_df["Silhouette"].idxmax(), "K"]
print("\nBest K by silhouette score:", best_k)
```
